```python
import statistics as st
import math as m


def summarize(data):
    n = len(data)
    mean = st.mean(data)
    sd = st.stdev(data)
    mn = min(data)
    mx = max(data)
    return n, mean, sd, mn, mx

data = [72, 85, 60, 91, 78, 55, 88, 74, 95, 63]

n, mean, sd, lo, hi = summarize(data)


print("N:", n)
print("Mean:", mean)
print("SD:", sd)
print("Min:", lo)
print("Max:", hi)
print("Range:", hi - lo)
```
